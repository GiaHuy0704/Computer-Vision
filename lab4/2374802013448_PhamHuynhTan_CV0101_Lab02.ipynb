# 5. Có nhận xét gì về các kích thước filter.

Dựa trên kết quả trực quan từ việc áp dụng các bộ lọc Box, Gaussian và Median, dưới đây là những nhận xét của tôi về tác động của các kích thước bộ lọc:

*   **Bộ lọc Box (Bộ lọc trung bình)**: Việc tăng kích thước bộ lọc (ví dụ: từ 3x3 lên 9x9) dẫn đến hiệu ứng làm mờ rõ rệt hơn và mất nhiều chi tiết hình ảnh hơn. Mặc dù các kernel lớn hơn giảm nhiễu hiệu quả hơn, chúng cũng làm cho hình ảnh tổng thể trở nên mờ hơn.
*   **Bộ lọc Gaussian**: Tương tự như bộ lọc Box, kích thước kernel Gaussian lớn hơn dẫn đến hiệu ứng làm mờ mạnh hơn và mất nhiều chi tiết hơn. Hiệu ứng làm mờ thường mượt mà và tự nhiên hơn so với bộ lọc Box, nhưng sự đánh đổi giữa giảm nhiễu và bảo toàn chi tiết vẫn tồn tại.
*   **Bộ lọc Median**: Đối với nhiễu hạt tiêu (pepper noise), bộ lọc Median rất hiệu quả. Với các kích thước kernel lớn hơn (ví dụ: 3, 5, 9), nó tiếp tục loại bỏ nhiễu hiệu quả. Tuy nhiên, các kích thước kernel rất lớn có thể dẫn đến làm phẳng đáng kể các kết cấu và chi tiết nhỏ, làm cho hình ảnh trông hơi 'sơn dầu' hoặc 'posterized', mặc dù nó vẫn vượt trội trong việc loại bỏ các điểm nhiễu biệt lập. Quan sát chung là bộ lọc Median giữ được các cạnh tốt hơn so với bộ lọc Box và Gaussian, đặc biệt ở các kích thước kernel nhỏ hơn.

Tóm lại, mặc dù các kích thước bộ lọc lớn hơn thường cải thiện khả năng giảm nhiễu trên cả ba loại, chúng cũng có xu hướng gây mất nhiều chi tiết hình ảnh hơn. Bộ lọc Median nổi bật nhờ khả năng loại bỏ nhiễu hạt tiêu đồng thời tương đối bảo toàn các cạnh, mặc dù các kích thước kernel rất lớn vẫn có thể ảnh hưởng đến các kết cấu tinh tế.

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Đọc ảnh
img1 = cv2.imread('exT3_01.jpg')
img2 = cv2.imread('exT3_02.jpg')

# Cắt ảnh
img1 = img1[:155, 1:191]
img2 = img2[:155, :190]

# Grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Lấy hiệu
diff = cv2.absdiff(gray1, gray2)

# Threshold
_, thresh = cv2.threshold(diff, 30, 255, cv2.THRESH_BINARY)

# Morphology
kernel = np.ones((5,5), np.uint8)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

# Mask lấy quả bóng
result = cv2.bitwise_and(img1, img1, mask=thresh)

# Hiển thị
print("Ảnh difference")
cv2_imshow(diff)

print("Mask quả bóng")
cv2_imshow(thresh)

print("Kết quả chỉ còn quả bóng")
cv2_imshow(result)
```

# BÀI TẬP MỞ RỘNG

```python
img_bgr = cv2.imread('cat.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Display the image to confirm successful loading and conversion
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title('Loaded Image (RGB)')
plt.axis('off')
plt.show()
```

```python
gray_img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

backward_diff_x = np.zeros_like(gray_img, dtype=np.float32)
backward_diff_x[:, 1:] = gray_img[:, 1:].astype(np.float32) - gray_img[:, :-1].astype(np.float32)

backward_diff_y = np.zeros_like(gray_img, dtype=np.float32)
backward_diff_y[1:, :] = gray_img[1:, :].astype(np.float32) - gray_img[:-1, :].astype(np.float32)

forward_diff_x = np.zeros_like(gray_img, dtype=np.float32)
forward_diff_x[:, :-1] = gray_img[:, 1:].astype(np.float32) - gray_img[:, :-1].astype(np.float32)

forward_diff_y = np.zeros_like(gray_img, dtype=np.float32)
forward_diff_y[:-1, :] = gray_img[1:, :].astype(np.float32) - gray_img[:-1, :].astype(np.float32)

central_diff_x = np.zeros_like(gray_img, dtype=np.float32)
central_diff_x[:, 1:-1] = (gray_img[:, 2:].astype(np.float32) - gray_img[:, :-2].astype(np.float32)) / 2.0

central_diff_y = np.zeros_like(gray_img, dtype=np.float32)
central_diff_y[1:-1, :] = (gray_img[2:, :].astype(np.float32) - gray_img[:-2, :].astype(np.float32)) / 2.0

backward_gradient_magnitude = np.sqrt(backward_diff_x**2 + backward_diff_y**2)
forward_gradient_magnitude = np.sqrt(forward_diff_x**2 + forward_diff_y**2)
central_gradient_magnitude = np.sqrt(central_diff_x**2 + central_diff_y**2)

plt.figure(figsize=(18, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray_img, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(backward_gradient_magnitude, cmap='gray')
plt.title('Backward Gradient Magnitude')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(forward_gradient_magnitude, cmap='gray')
plt.title('Forward Gradient Magnitude')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(central_gradient_magnitude, cmap='gray')
plt.title('Central Gradient Magnitude')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

finite_diff_x_component = np.zeros_like(gray_img, dtype=np.float32)
finite_diff_x_component[:, 1:] = np.abs(gray_img[:, 1:].astype(np.float32) - gray_img[:, :-1].astype(np.float32))

finite_diff_y_component = np.zeros_like(gray_img, dtype=np.float32)
finite_diff_y_component[1:, :] = np.abs(gray_img[1:, :].astype(np.float32) - gray_img[:-1, :].astype(np.float32))

finite_difference_image = np.sqrt(finite_diff_x_component**2 + finite_diff_y_component**2)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(finite_difference_image, cmap='gray')
plt.title('Finite Difference Filter (Magnitude)')
plt.axis('off')

plt.tight_layout()
plt.show()

print("Finite Difference Filter applied and displayed.")
```

```python
gray_img_sobel = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Apply Sobel operator in x-direction
sobelx = cv2.Sobel(gray_img_sobel, cv2.CV_64F, 1, 0, ksize=3)

# Apply Sobel operator in y-direction
sobely = cv2.Sobel(gray_img_sobel, cv2.CV_64F, 0, 1, ksize=3)

# Convert absolute values to 8-bit unsigned integers
abs_sobelx = cv2.convertScaleAbs(sobelx)
abs_sobely = cv2.convertScaleAbs(sobely)

# Combine the x and y gradients
sobel_combined = cv2.addWeighted(abs_sobelx, 0.5, abs_sobely, 0.5, 0)

# Display the results
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray_img_sobel, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(abs_sobelx, cmap='gray')
plt.title('Sobel X')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(abs_sobely, cmap='gray')
plt.title('Sobel Y')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Combined')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
gray_img_canny = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

edges_low_thresh = cv2.Canny(gray_img_canny, 50, 150)
edges_mid_thresh = cv2.Canny(gray_img_canny, 100, 200)
edges_high_thresh = cv2.Canny(gray_img_canny, 150, 250)

plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray_img_canny, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(edges_low_thresh, cmap='gray')
plt.title('Canny (Low Thresh)')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(edges_mid_thresh, cmap='gray')
plt.title('Canny (Mid Thresh)')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(edges_high_thresh, cmap='gray')
plt.title('Canny (High Thresh)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np

edges_for_hough = edges_mid_thresh

lines = cv2.HoughLinesP(edges_for_hough, 1, np.pi / 180, 100, minLineLength=100, maxLineGap=10)

line_image = np.copy(img_rgb) * 0

if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]
        cv2.line(line_image, (x1, y1), (x2, y2), (255, 0, 0), 5)

lines_on_original = cv2.addWeighted(img_rgb, 0.8, line_image, 1, 0)

plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(edges_for_hough, cmap='gray')
plt.title('Canny Edges for Hough')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(lines_on_original)
plt.title('Hough Lines Detected')
plt.axis('off')

plt.tight_layout()
plt.show()
```
