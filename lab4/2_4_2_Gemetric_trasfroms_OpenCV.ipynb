Ô này nhập các thư viện cần thiết: `matplotlib.pyplot` (được đặt biệt danh là `plt`) để vẽ đồ thị, `cv2` (OpenCV) để xử lý hình ảnh và `numpy` (được đặt biệt danh là `np`) cho các phép toán số học.

```python
def plot_image(image_1, image_2,title_1="Orignal",title_2="New Image"):
    plt.figure(figsize=(10,10))
    plt.subplot(1, 2, 1)
    plt.imshow(image_1,cmap="gray")
    plt.title(title_1)
    plt.subplot(1, 2, 2)
    plt.imshow(image_2,cmap="gray")
    plt.title(title_2)
    plt.show()
```

Mã này định nghĩa một hàm trợ giúp `plot_image` để hiển thị hai hình ảnh cạnh nhau. Nó tạo một hình với hai ô phụ, hiển thị `image_1` trong ô đầu tiên và `image_2` trong ô thứ hai, có thể tùy chỉnh tiêu đề và sử dụng bản đồ màu xám.

```python
toy_image = np.zeros((6,6))
toy_image[1:5,1:5]=255
toy_image[2:4,2:4]=0
plt.imshow(toy_image,cmap='gray')
plt.show()
toy_image
```

Mã này tạo một 'hình ảnh đồ chơi' dạng ma trận NumPy 6x6. Nó khởi tạo tất cả các pixel bằng 0 (đen), sau đó tạo một hình vuông màu trắng (255) bên trong và một hình vuông màu đen (0) nhỏ hơn bên trong hình vuông màu trắng. Cuối cùng, nó hiển thị hình ảnh này bằng cách sử dụng `matplotlib` với bản đồ màu xám và in ra ma trận NumPy của hình ảnh.

```python
new_toy = cv2.resize(toy_image,None,fx=2, fy=1, interpolation = cv2.INTER_NEAREST )
plt.imshow(new_toy,cmap='gray')
plt.show()
```

Mã này thay đổi kích thước `toy_image` bằng cách sử dụng `cv2.resize`. Nó phóng to hình ảnh theo chiều ngang (yếu tố tỷ lệ `fx=2`) và giữ nguyên chiều dọc (`fy=1`). `interpolation=cv2.INTER_NEAREST` sử dụng phương pháp nội suy pixel gần nhất. Hình ảnh mới sau đó được hiển thị.

```python
image = cv2.imread("lenna.png")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.show()
```

Mã này đọc tệp hình ảnh 'lenna.png' bằng `cv2.imread()`. Sau đó, nó chuyển đổi hình ảnh từ định dạng BGR (mặc định của OpenCV) sang RGB (định dạng mong muốn của Matplotlib) bằng `cv2.cvtColor()` để đảm bảo màu sắc được hiển thị chính xác. Cuối cùng, nó hiển thị hình ảnh đã chuyển đổi.

```python
new_image = cv2.resize(image, None, fx=2, fy=1, interpolation=cv2.INTER_CUBIC)
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape)
```

Mã này thay đổi kích thước hình ảnh theo chiều ngang bằng cách sử dụng `cv2.resize()`. `fx=2` có nghĩa là chiều rộng được nhân đôi, trong khi `fy=1` giữ nguyên chiều cao. `cv2.INTER_CUBIC` được sử dụng để nội suy pixel. Sau đó, hình ảnh mới được hiển thị và kích thước của cả hình ảnh cũ và mới được in ra.

```python
new_image = cv2.resize(image, None, fx=1, fy=2, interpolation=cv2.INTER_CUBIC)
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape)
```

Mã này thay đổi kích thước hình ảnh theo chiều dọc bằng cách sử dụng `cv2.resize()`. `fx=1` giữ nguyên chiều rộng, trong khi `fy=2` có nghĩa là chiều cao được nhân đôi. `cv2.INTER_CUBIC` được sử dụng để nội suy pixel. Sau đó, hình ảnh mới được hiển thị và kích thước của cả hình ảnh cũ và mới được in ra.

Mã này thay đổi kích thước hình ảnh theo cả chiều ngang và chiều dọc, nhân đôi cả chiều rộng (`fx=2`) và chiều cao (`fy=2`). `cv2.INTER_CUBIC` được sử dụng để nội suy. Sau đó, hình ảnh đã phóng to được hiển thị và kích thước của hình ảnh gốc và hình ảnh mới được in ra để so sánh.

```python
new_image = cv2.resize(image, None, fx=1, fy=0.5, interpolation=cv2.INTER_CUBIC)
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape)
```

Mã này thu nhỏ hình ảnh theo chiều dọc bằng cách sử dụng `cv2.resize()`. `fx=1` giữ nguyên chiều rộng, trong khi `fy=0.5` giảm một nửa chiều cao. `cv2.INTER_CUBIC` được sử dụng để nội suy pixel. Sau đó, hình ảnh mới được hiển thị và kích thước của cả hình ảnh cũ và mới được in ra.

```python
rows = 100
cols = 200
```

Mã này định nghĩa hai biến số nguyên: `rows` được gán giá trị 100 và `cols` được gán giá trị 200. Các biến này có thể được sử dụng để xác định kích thước (chiều cao và chiều rộng) của hình ảnh hoặc ma trận.

```python
new_image = cv2.resize(image, (100, 200), interpolation=cv2.INTER_CUBIC)
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape)
```

Mã này thay đổi kích thước `image` thành kích thước cố định là 100 pixel chiều rộng và 200 pixel chiều cao bằng cách sử dụng `cv2.resize()`. `interpolation=cv2.INTER_CUBIC` được sử dụng để nội suy pixel. Sau đó, hình ảnh đã thay đổi kích thước được hiển thị và kích thước của hình ảnh gốc và hình ảnh mới được in ra.

```python
tx = 100
ty = 0
M = np.float32([[1, 0, tx], [0, 1, ty]])
M
```

Mã này định nghĩa các giá trị dịch chuyển `tx` (theo chiều ngang) và `ty` (theo chiều dọc) là 100 và 0 tương ứng. Sau đó, nó tạo một ma trận chuyển đổi 2D `M` bằng cách sử dụng NumPy, đây là ma trận nhận dạng với các giá trị dịch chuyển được thêm vào hàng cuối cùng của nó. Ma trận `M` được in ra.

```python
rows, cols, _ = image.shape
```

Mã này trích xuất chiều cao (`rows`) và chiều rộng (`cols`) của `image` bằng cách giải nén thuộc tính `shape` của nó. Giá trị thứ ba (số kênh màu) được bỏ qua bằng cách sử dụng `_`.

```python
new_image = cv2.warpAffine(image, M, (cols, rows))
```

Mã này áp dụng phép biến đổi affine (dịch chuyển) cho `image` bằng cách sử dụng ma trận `M` đã định nghĩa trước đó. Hàm `cv2.warpAffine()` thực hiện phép biến đổi. Kích thước của hình ảnh đầu ra được đặt thành kích thước ban đầu của hình ảnh (`cols`, `rows`), có thể dẫn đến việc cắt xén nếu hình ảnh được dịch chuyển ra ngoài ranh giới.

```python
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.show()
```

Mã này hiển thị `new_image` (hình ảnh đã được dịch chuyển). Trước khi hiển thị, nó chuyển đổi hình ảnh từ định dạng BGR sang RGB bằng `cv2.cvtColor()` để đảm bảo màu sắc chính xác với `matplotlib.pyplot`.

```python
new_image = cv2.warpAffine(image, M, (cols + tx, rows + ty))
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.show()
```

Mã này áp dụng phép biến đổi affine (dịch chuyển) tương tự cho `image` như trước, nhưng lần này nó điều chỉnh kích thước hình ảnh đầu ra thành `(cols + tx, rows + ty)`. Việc điều chỉnh này đảm bảo rằng toàn bộ hình ảnh đã dịch chuyển được hiển thị mà không bị cắt xén. Hình ảnh kết quả sau đó được hiển thị.

```python
tx = 0
ty = 50
M = np.float32([[1, 0, tx], [0, 1, ty]])
new_iamge = cv2.warpAffine(image, M, (cols + tx, rows + ty))
plt.imshow(cv2.cvtColor(new_iamge, cv2.COLOR_BGR2RGB))
plt.show()
```

Mã này thực hiện dịch chuyển hình ảnh theo chiều dọc. Nó đặt `tx` (dịch chuyển ngang) bằng 0 và `ty` (dịch chuyển dọc) bằng 50. Sau đó, nó tạo một ma trận biến đổi `M` mới cho phép dịch chuyển này, áp dụng ma trận cho hình ảnh bằng `cv2.warpAffine()` với kích thước đầu ra được điều chỉnh, và cuối cùng hiển thị hình ảnh đã dịch chuyển.

```python
theta = 45.0
M = cv2.getRotationMatrix2D(center=(3, 3), angle=theta, scale=1)
new_toy_image = cv2.warpAffine(toy_image, M, (6, 6))
```

Mã này đặt góc xoay `theta` là 45 độ. Nó sử dụng `cv2.getRotationMatrix2D()` để tạo một ma trận xoay `M` xoay hình ảnh quanh tâm `(3, 3)` của `toy_image` mà không thay đổi tỷ lệ (`scale=1`). Cuối cùng, nó áp dụng ma trận xoay này cho `toy_image` bằng `cv2.warpAffine()` để tạo `new_toy_image` đã xoay.

```python
plot_image(toy_image, new_toy_image, title_1="Orignal", title_2="rotated image")
```

Mã này sử dụng hàm trợ giúp `plot_image` đã định nghĩa trước đó để hiển thị `toy_image` gốc và `new_toy_image` (hình ảnh đồ chơi đã xoay) cạnh nhau. Nó đặt tiêu đề cho hình ảnh gốc là 'Original' và hình ảnh đã xoay là 'rotated image'.

Cuối cùng, mảng `Noise` được chuyển đổi sang kiểu dữ liệu `np.uint8` và kích thước của nó được in ra.

```python
im_gray = cv2.imread('barbara.png', cv2.IMREAD_GRAYSCALE)
im_gray.shape

plt.imshow(im_gray,cmap='gray')
plt.show()
```

Mã này đọc tệp hình ảnh 'barbara.png' dưới dạng hình ảnh xám bằng cách sử dụng `cv2.imread()` với cờ `cv2.IMREAD_GRAYSCALE`. Nó in ra kích thước của hình ảnh xám, sau đó hiển thị hình ảnh bằng `matplotlib` với bản đồ màu xám.

```python
U, s, V = np.linalg.svd(im_gray , full_matrices=True)
```

Mã này thực hiện phép phân tích giá trị kỳ dị (SVD) trên `im_gray` (hình ảnh xám). Nó phân tách ma trận hình ảnh thành ba ma trận: `U` (ma trận đơn vị), `s` (mảng các giá trị kỳ dị), và `V` (ma trận đơn vị). `full_matrices=True` đảm bảo `U` và `V` là ma trận vuông.

```python
s.shape
```

Mã này in ra kích thước của mảng `s`, chứa các giá trị kỳ dị từ phép phân tích SVD. Điều này xác nhận rằng `s` là một mảng một chiều (vector).

```python
S = np.zeros((im_gray.shape[0], im_gray.shape[1]))
S[:image.shape[0], :image.shape[0]] = np.diag(s)
```

Mã này tạo một ma trận không mới `S` với cùng kích thước như `im_gray`. Sau đó, nó điền các giá trị kỳ dị từ mảng `s` vào đường chéo chính của `S`, biến đổi các giá trị kỳ dị 1D thành một ma trận đường chéo `S` có kích thước đầy đủ.

```python
 plot_image(U,V,title_1="Matrix U ",title_2="matrix  V")
```

Mã này sử dụng hàm `plot_image` đã định nghĩa trước đó để hiển thị hai ma trận `U` và `V` (từ phép phân tích SVD) cạnh nhau. Nó đặt tiêu đề 'Matrix U' cho ma trận `U` và 'matrix V' cho ma trận `V`.

```python
plt.imshow(S,cmap='gray')
plt.show()
```

Mã này hiển thị ma trận `S` (ma trận đường chéo chứa các giá trị kỳ dị) bằng `matplotlib` với bản đồ màu xám. Điều này trực quan hóa phân bố của các giá trị kỳ dị, thường cho thấy hầu hết các giá trị đều bằng không ngoại trừ các phần tử trên đường chéo chính.

```python
B = S.dot(V)
plt.imshow(B,cmap='gray')
plt.show()
```

Mã này thực hiện phép nhân ma trận giữa ma trận `S` (đường chéo các giá trị kỳ dị) và ma trận `V` (từ SVD). Kết quả được lưu trữ trong biến `B` và sau đó được hiển thị dưới dạng hình ảnh xám bằng `matplotlib`.

```python
A = U.dot(B)
```

Mã này thực hiện phép nhân ma trận giữa ma trận `U` (từ SVD) và ma trận `B` (là kết quả của `S.dot(V)`). Phép toán này tái tạo lại hình ảnh gốc `im_gray` (hoặc một xấp xỉ rất gần của nó) và lưu trữ kết quả trong biến `A`.

```python
plt.imshow(A,cmap='gray')
plt.show()
```

Mã này hiển thị hình ảnh `A` đã được tái tạo (là kết quả của `U.dot(S.dot(V))`) bằng cách sử dụng `matplotlib` với bản đồ màu xám. Hình ảnh này sẽ trông gần giống với hình ảnh xám gốc `im_gray`.
